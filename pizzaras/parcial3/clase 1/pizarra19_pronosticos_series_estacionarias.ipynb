# ***Pronósticos de series estacionarias***

Suponga que $ {W_t} $ es una serie de tiempo estacionaria  
con media cero obtenida a partir de un proceso $ \{Z_t\}_{t=1} N$ .  
Entonces:

$$
W_t = \nabla^d\, T(Z_t)
\tag{1}
$$

Para algún \(d\) y para una cierta transformación \(T\).

Recuerde que $ W_t $ admite la representación:

$$
W_t = \Phi(B)\, a_t,
\qquad a_t \sim WN(0,\sigma_a^2)
\tag{2}
$$

Para la cual existe un modelo ARIMA equivalente.

---




Si a partir del origen $ t $ se desea pronosticar la observación  
$ W_{t+h} $, un pronóstico cualquiera de esta observación, que  
se obtenga como combinación lineal de los valores de $\{W_t\}$  
y en consecuencia de $\{Z_t\}$, será denotado

$$
\tilde{W}_t(h)
$$

mientras que el pronóstico óptimo se denotará

$$
\hat{W}_t(h)
$$

El criterio que se usará para determinar la optimalidad  
del pronóstico será el de error cuadrático medio mínimo,  
es decir, $\hat{W}_t(h)$ satisface la condición:


La condición de optimalidad es:

$$
E_t \left[\, W_{t+h} - \hat{W}_t(h) \,\right]^2
\;=\;
\min_{\tilde{W}_t(h)}
E_t \left[\, W_{t+h} - \tilde{W}_t(h) \,\right]^2
\tag{3}
$$

en la cual $E_t$ denota la esperanza condicional dada la información  
hasta el tiempo $t$.

Es decir,

$$
E_t \left[\, W_{t+h} - \hat{W}_t(h) \,\right]^2
\;=\;
E \left\{\, 
\left[ W_{t+h} - \tilde{W}_t(h) \right]^2
\;\middle|\;
Z_t, Z_{t-1}, \ldots, Z_1
\right\}
\tag{4}
$$


El pronóstico con error cuadrático medio mínimo está dado por

$$
\hat{W}_t(h) = E_t(W_{t+h})
\tag{5}
$$

En este caso los errores de pronóstico están dados por

$$
e_t(h) = W_{t+h} - \hat{W}_t(h)
\tag{6}
$$


Note que

i) 
$$
W_{t+h}
= -\sum_{j=0}^{\infty} \psi_j\, a_{t+h-j}
$$

ii)
$$
\hat{W}_t(h)
= -\sum_{j=h}^{\infty} \psi_j\, a_{t+h-j}
$$



Luego,

$$
e_t(h)
= -\sum_{j=0}^{\infty} \psi_{t+h-j}\
-
\left( -\sum_{j=h}^{\infty} \psi_j\, a_{t+h-j} \right)
$$



$$
e_t(h)
=
\left[
-\sum_{j=0}^{h-1} \psi_j\, a_{t+h-j} - \sum_{j=h}^{\infty} \psi_j\, a_{t+h-j}
\right]
-
\left[
-
\sum_{j=h}^{\infty} \psi_j\, a_{t+h-j}
\right]
$$




Por lo tanto,

$$
e_t(h)
=
-\sum_{j=0}^{h-1} \psi_j\, a_{t+h-j}
\tag{7}
$$
