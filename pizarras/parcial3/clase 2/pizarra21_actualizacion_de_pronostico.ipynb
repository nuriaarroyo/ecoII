# ***Actualización del pronóstico***

Dado que las series de tiempo aumentan su número de observaciones conforme
aumenta el tiempo, esta situación afecta a los pronósticos.




Por ejemplo, el pronóstico $\widehat T(Z_t)(h)$ pierde validez al conocerse
la observación $T(Z_{t+1})$. Con la finalidad de actualizar pronósticos,
recordemos que:

i)
$$
\mathbb{E}_t(W_{t+h}) = \widehat{W}_t(h),
$$

ii)
$$
\widehat{W}_t(h) = - \sum_{j=h}^{\infty} \psi_j\, a_{t+h-j}.
$$

Por lo tanto,

$$
\mathbb{E}_t\!\big(T(Z_{t+h})\big)
=
\widehat T(Z_t)(h)
=
- \sum_{j=h}^{\infty} \psi_j\, a_{t+h-j}.
\qquad (1)
$$




De aquí se tiene que

$$
\widehat T(Z_{t+1})(h)
=
\widehat T(Z_t)(h+1)
-
\psi_h\,\hat{a}_{t+1},
\qquad h \ge 1.
\qquad (2)
$$



En efecto,

i)
$$
\widehat T(Z_{t+1})(h)
=
\mathbb{E}_{t+1}\!\big(T(Z_{t+1+h})\big)
=
- \sum_{j=h}^{\infty} \psi_j\, a_{t+1+h-j},
$$



ii)
$$
\widehat T(Z_t)(h+1)
=
\mathbb{E}_t\!\big(T(Z_{t+h+1})\big)
=
- \sum_{j=h}^{\infty} \psi_j\, a_{t+h+1-j}.
$$



Por lo tanto,

$$
\widehat T(Z_{t+1})(h)
=
- \sum_{j=h+1}^{\infty} \psi_j\, a_{t+h+1-j}
-
\psi_h\, a_{t+1}.
$$

La expresión (2) sirve para **actualizar los pronósticos** una vez que se ha
observado un dato más en la serie.


**Ejemplo.** Considérese la serie de CETES del ejemplo de **pronósticos e
intervalos de predicción**.


i)
$$
\widehat T(\text{CETES}_{t=57})(1) = 7696.85
$$

ii)
$$
\widehat T(\text{CETES}_{t=58})(2) = 7941.57
$$

iii)
$$
\widehat T(\text{CETES}_{t=59})(3) = 7709.88
$$

$$
\text{CETES}_{t=60},
$$

información disponible hasta $t = 60$. Por lo tanto, es necesario
**actualizar el pronóstico calculado anteriormente**.

Recordemos que la actualización se realiza con la expresión (2).

El cálculo del residual $\hat{a}_{t+1}$se puede realizar mediante
**dos formas**:

**1)**

$$
\hat{a}_{60}
=
T(\text{CETES}_{t=60})
-
\widehat T(\text{CETES}_{t=59})(1).
$$

En general,

$$
\hat{a}_{t+1}
=
T(\text{CETES}_{t+1})
-
\widehat T(\text{CETES}_{t})(1).
$$



**2)** Usando el método de **residuales aproximados**.

En este método, se busca una expresión para los residuales en el tiempo
$t$ y se aproximan de forma recursiva.



Por ejemplo, en la serie de CETES, el modelo considerado fue

$$
(1 - 0.388\,B^4)\,\nabla T(\text{CETES}_t)
=
(1 - 0.481\,B^2)\,\hat{a}_t.
$$

De donde

$$
\nabla T(\text{CETES}_t)
-
0.388\,\nabla T(\text{CETES}_{t-4})
=
\hat{a}_t
-
0.481\,\hat{a}_{t-2}.
$$

Luego,

$$
\hat{a}_t
=
\nabla T(\text{CETES}_t)
-
0.388\,\nabla T(\text{CETES}_{t-4})
+
0.481\,\hat{a}_{t-2}.
$$

Considérese
$$
\hat{a}_{t=18} = \hat{a}_{t=19} = 0.
$$

A partir de estos valores iniciales, se puede calcular de forma recursiva
$\hat{a}_0, \hat{a}_1, \ldots, \hat{a}_t$.

En particular,

$$
\hat{a}_{20}
=
\nabla T(\text{CETES}_{t=20})
-
0.388\,\nabla T(\text{CETES}_{t=16})
+
0.481\,\hat{a}_{18}.
$$
