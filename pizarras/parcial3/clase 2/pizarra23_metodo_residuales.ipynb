# ***Método de aproximación de residuales***

Considérese la serie de $CETES_t$, cuyo modelo está dado por

$$
(1 - 0.388\,B^4)\,\nabla T(\text{CETES}_t)
=
(1 - 0.481\,B^2)\,\hat{a}_t,
$$

con

$$
\hat{\sigma}_a = 1107.7,
\qquad
T(\text{CETES}_t) = \frac{\text{CETES}_t}{1000}.
$$

**Desarrollando**, se tiene

$$
\nabla T(\text{CETES}_t)
-
0.388\,\nabla T(\text{CETES}_{t-4})
=
\hat{a}_t
-
0.481\,\hat{a}_{t-2}.
$$

Esto es,

$$
T(\text{CETES}_t)
-
T(\text{CETES}_{t-1})
-
0.388\bigl(
T(\text{CETES}_{t-4})
-
T(\text{CETES}_{t-5})
\bigr)
=
\hat{a}_t
-
0.481\,\hat{a}_{t-2}.
$$

Por lo tanto,

$$
\hat{a}_t
=
T(\text{CETES}_t)
-
T(\text{CETES}_{t-1})
-
0.388\,T(\text{CETES}_{t-4})
+
0.388\,T(\text{CETES}_{t-5})
+
0.481\,\hat{a}_{t-2}.
$$

Considérese como valores iniciales

$$
\hat{a}_{18} = 0,
\qquad
\hat{a}_{19} = 0.
$$

Después de aplicar el método de **aproximación de residuales**, se obtiene

$$
\hat{a}_{60} = -799,
$$

y con este valor calculamos la **actualización del pronóstico**:



$$
\widehat T(\text{CETES}_{t=60})(1)
=
\widehat T(\text{CETES}_{t=57})(2)
-
\hat{\psi}_1\,\hat{a}_{60}.
$$

Como \(\hat{\psi}_1 = -1\),

$$
\widehat T(\text{CETES}_{t=60})(1)
=
7941.57
-
(-1)(-799).
$$

Así,

$$
\widehat T(\text{CETES}_{t=60})(1)
=
7941.57 - 799
=
7141.16.
$$
