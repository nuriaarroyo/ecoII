# ***Pronóstico de la serie original***

Supóngase que $\{Z_t\}$ es un proceso que genera a una serie de tiempo y ha sido
transformada mediante $T(Z_t)$.

**i) Caso 1: $T$ es una transformación lineal**

En este caso, $\widehat{Z}_t(h)$, el pronóstico óptimo de la serie original, será

$$
\widehat{Z}_t(h)
=
T^{-1}\!\left[\,\widehat T(Z_t)(h)\,\right].
$$

**ii) Caso 2: $T$ no es lineal**

1) Supóngase que $T$ es la transformación de **Box–Cox**, es decir,

$$
Z_t^{\,\lambda}
=
\begin{cases}
\dfrac{Z_t^{\lambda}-1}{\lambda}, & \lambda \neq 0, \\[10pt]
\log(Z_t), & \lambda = 0.
\end{cases}
$$

Entonces, se usará la siguiente **corrección de sesgo**.


$$
\widehat{C}_{t,\lambda}(h)
=
\begin{cases}
\left\{
\dfrac{1}{2}
+
\sqrt{
1-{2}
+
\lambda(\lambda-1)
\left[
1
+
\lambda\,\widehat T(Z_t)(h)
\right]^{-2}
{Var}(e_t(h))/2
}
\right\}^{1/\lambda},
& \lambda \neq 0,
\\[14pt]

\exp\!\left[
\dfrac{1}{2}\,
\widehat{\operatorname{Var}}_t\!\big(e_t(h)\big)
\right],
& \lambda = 0.
\end{cases}
$$

donde \(\widehat{\operatorname{Var}}_t\!\big(e_t(h)\big)\) es el estimador de
\(\operatorname{Var}_t\!\big(e_t(h)\big)\).

Es decir,

$$
\widehat{\operatorname{Var}}_t\!\big(e_t(h)\big)
=
\sum_{j=0}^{h-1}
\psi_j^{\,2}\,
\hat{\sigma}_a^{\,2}.
$$


Finalmente, el pronóstico óptimo de la serie original está dado por

$$
\widehat{Z}_t(h)
=
\mathbb{E}_t\!\left\{
T^{-1}\!\big( T(Z_t)(h) \big)
\right\}
$$

$$
=
T^{-1}\!\left\{
\widehat{T(Z_t)}(h)
\right\}
=
\widehat{C}_{t,\lambda}(h).
$$

Mientras que el **intervalo de predicción** al nivel \((1-\alpha)100\%\)
de probabilidad será

$$
\widehat{Z}_{t,\lambda}(h)
\in
T^{-1}\!\left\{
\widehat{T(Z_t)}(h)
\pm
z_{\alpha/2}
\sqrt{
\widehat{\operatorname{Var}}_t\!\big(e_t(h)\big)
}
\right\}.
$$
