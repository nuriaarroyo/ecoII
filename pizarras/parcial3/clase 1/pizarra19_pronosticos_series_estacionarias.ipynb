# ***Pronósticos de series estacionarias***

Suponga que $ {W_t} $ es una serie de tiempo estacionaria  
con media cero obtenida a partir de un proceso $ \{Z_t\}_{t=1} N$ .  
Entonces:

$$
W_t = \nabla^d\, T(Z_t)
\tag{1}
$$

Para algún \(d\) y para una cierta transformación \(T\).

Recuerde que $ W_t $ admite la representación:

$$
W_t = \Phi(B)\, a_t,
\qquad a_t \sim WN(0,\sigma_a^2)
\tag{2}
$$

Para la cual existe un modelo ARIMA equivalente.

---




Si a partir del origen $ t $ se desea pronosticar la observación  
$ W_{t+h} $, un pronóstico cualquiera de esta observación, que  
se obtenga como combinación lineal de los valores de $\{W_t\}$  
y en consecuencia de $\{Z_t\}$, será denotado

$$
\tilde{W}_t(h)
$$

mientras que el pronóstico óptimo se denotará

$$
\hat{W}_t(h)
$$

El criterio que se usará para determinar la optimalidad  
del pronóstico será el de error cuadrático medio mínimo,  
es decir, $\hat{W}_t(h)$ satisface la condición:


La condición de optimalidad es:

$$
E_t \left[\, W_{t+h} - \hat{W}_t(h) \,\right]^2
\;=\;
\min_{\tilde{W}_t(h)}
E_t \left[\, W_{t+h} - \tilde{W}_t(h) \,\right]^2
\tag{3}
$$

en la cual $E_t$ denota la esperanza condicional dada la información  
hasta el tiempo $t$.

Es decir,

$$
E_t \left[\, W_{t+h} - \hat{W}_t(h) \,\right]^2
\;=\;
E \left\{\, 
\left[ W_{t+h} - \tilde{W}_t(h) \right]^2
\;\middle|\;
Z_t, Z_{t-1}, \ldots, Z_1
\right\}
\tag{4}
$$


El pronóstico con error cuadrático medio mínimo está dado por

$$
\hat{W}_t(h) = E_t(W_{t+h})
\tag{5}
$$

En este caso los errores de pronóstico están dados por

$$
e_t(h) = W_{t+h} - \hat{W}_t(h)
\tag{6}
$$


Note que

i) 
$$
W_{t+h}
= -\sum_{j=0}^{\infty} \psi_j\, a_{t+h-j}
$$

ii)
$$
\hat{W}_t(h)
= -\sum_{j=h}^{\infty} \psi_j\, a_{t+h-j}
$$



Luego,

$$
e_t(h)
= -\sum_{j=0}^{\infty} \psi_{t+h-j}\
-
\left( -\sum_{j=h}^{\infty} \psi_j\, a_{t+h-j} \right)
$$



$$
e_t(h)
=
\left[
-\sum_{j=0}^{h-1} \psi_j\, a_{t+h-j} - \sum_{j=h}^{\infty} \psi_j\, a_{t+h-j}
\right]
-
\left[
-
\sum_{j=h}^{\infty} \psi_j\, a_{t+h-j}
\right]
$$




Por lo tanto,

$$
e_t(h)
=
-\sum_{j=0}^{h-1} \psi_j\, a_{t+h-j}
\tag{7}
$$


---


Para otra parte, ya que se desea calcular el pronóstico de ( Y_{t+h} ), considérese un modelo **ARIMA((p,d,q))**:

$$
\phi(B) W_t = \theta(B) a_t, \qquad a_t \sim \text{WN}(0,\sigma_a^2)
\tag{8}
$$



en donde

$$
W_t = \nabla^d Z_t.
$$

Por la ecuación (5), se tiene que

$$
\widehat{W}_t(h) = \mathbb{E}_t\left(W_{t+h}\right).
$$



De (8) tendremos que

$$
(1 - \phi_1 B - \phi_2 B^2 - \cdots - \phi_p B^p) W_t =
(1 - \theta_1 B - \theta_2 B^2 - \cdots - \theta_q B^q) a_t.
$$

Equivalentemente,

$$
W_t = 

\phi_1 W_{t-1} + \phi_2 W_{t-2} + \cdots + \phi_p W_{t-p}
 - a_t - \theta_1 a_{t-1} - \cdots - \theta_q a_{t-q}.
  $$



$$
\widehat{W}_t(h)
=
\mathbb{E}_t\!\Big(
\phi_1 W_{t+h-1}
+ \phi_2 W_{t+h-2}
+ \cdots
+ \phi_p W_{t+h-p}
- a_{t+h}
- \theta_1 a_{t+h-1}
- \cdots
- \theta_q a_{t+h-q}
\Big)
$$




$$
=
\phi_1 \mathbb{E}_t(W_{t+h-1})
+ \cdots
+ \phi_p \mathbb{E}_t(W_{t+h-p})
- \mathbb{E}_t(a_{t+h})
- \theta_1 \mathbb{E}_t(a_{t+h-1})
- \cdots
- \theta_q \mathbb{E}_t(a_{t+h-q})
$$

$$
\mathbb{E}_t(W_{t+h-j}) =
\begin{cases}
W_{t+h-j}, & j \ge h, \\[6pt]
\widehat{W}_t(h-j), & j < h,
\end{cases}
\qquad \text{(9)}
$$





$$
\mathbb{E}_t(a_{t+h-j}) =
\begin{cases}
W_{t+h-j}-\widehat{W}_{t+h-j-1}, & j \ge h, \\[6pt]
0, & j < h,
\end{cases}
\qquad (10)
$$


De aquí que,
$$
\mathbb{E}_t(\widehat{a}_{t+h-j}) =
\begin{cases}
\widehat{a}_{t+h-j}, & j \ge h, \\[6pt]
0, & j < h,
\end{cases}
\qquad (11)
$$


**Ejemplo.** Supóngase que el comportamiento de $W_t$ está dado por el proceso

$$
(1-0.6B)W_t = (1+0.2B)a_t
\qquad \text{ARIMA}(1,d,1)
$$

Entonces,

$$
W_t = 0.6\,W_{t-1} + a_t + 0.2\,a_{t-1}
$$

De tal forma que el pronóstico óptimo en $h=1$ es

$$
\widehat{W}_t(h) = \mathbb{E}_t(W_{t+h})
\tag{5}
$$


$$
\widehat{W}_t(1) = \mathbb{E}_t(W_{t+1})
$$



$$
\widehat{W}_t(1)
=
0.6\,\mathbb{E}_t(W_{t+1-1})
+ \mathbb{E}_t(a_{t+1})
+ 0.2\,\mathbb{E}_t(a_{t+1-1})
$$

$$
\widehat{W}_t(1)
\qquad =\qquad 
\underbrace{\mathbb{E}_t(W_t)}_{\ast}
\qquad +\qquad 
\underbrace{\mathbb{E}_t(a_{t+1})}_{\ast\ast}
\qquad +\qquad 
\underbrace{\mathbb{E}_t(a_t)}_{\ast\ast\ast}
$$

$$
\ast \quad \mathbb{E}_t(W_t) = W_t \qquad \qquad   h - j = 0 \;\Rightarrow\; h = j
$$

$$
\ast\ast \quad \mathbb{E}_t(a_{t+1}) = 0 \qquad \qquad  h - j = 1 \;\Rightarrow\; j < h
$$

$$
\ast\ast\ast \quad \mathbb{E}_t(a_t)
= W_t - \widehat{W}_{t-1}(1) \qquad \qquad h - j = 0 \;\Rightarrow\; h = j
$$



Por lo tanto,

$$
\widehat{W}_t(h=1)
=
0.6\,W_t
+ 0.2\,[\,W_t - \widehat{W}_{t-1}(1)\,].
$$




Ahora calculemos $\widehat{W}_t(h=2)$.

$$
\widehat{W}_t(h=2)
=
\mathbb{E}_t(W_{t+2})
=
0.6\,\widehat{W}_t(1).
$$

$$
\vdots
$$

En general,

$$
\widehat{W}_t(h)
=
0.6\,\widehat{W}_t(h-1).
$$

$$
\widehat{W}_t(1) \Rightarrow \widehat{W}_0(1),
\qquad \text{como }\qquad\widehat{W}_0(1)=W_1.
$$
