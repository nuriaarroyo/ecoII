# ***Intervalos de predicción***


Sea $\{Z_t\}$ el proceso que genera una serie de tiempo. Entonces, dicha serie puede ser modelada de acuerdo al siguiente **modelo ARIMA**:



$$
\Phi(B)\,T(Z_t) = \Theta(B)\,a_t
\qquad (1)
$$



donde

$$
a_t \sim \text{WN}(0,\sigma_a^2),
\qquad
\Phi(B) = \nabla^d \,\phi(B).
$$



Entonces,

$$
T(Z_t)
=
\Phi^{-1}(B)\,\Theta(B)\,a_t
=
\Psi(B)\,a_t
\qquad (2)
$$



con

$$
\Psi(B) = \Phi^{-1}(B)\,\Theta(B),
$$

de donde se obtiene que

$$
\Phi(B)\,\Psi(B) = \Theta(B).
\tag (3)
$$


---



Supóngase que \(d = 1\). Entonces,

$
\Phi(B) = \nabla\,\phi(B),
$

es decir,

$$
\Phi(B)
=
(1 - B)\bigl(1 - \phi_1 B - \phi_2 B^2 - \cdots - \phi_p B^p\bigr).
$$

Desarrollando, se obtiene

$$
\Phi(B)
=
1 - \phi_1 B - \phi_2 B^2 - \cdots - \phi_p B^p
- \phi_{p+1} B^{p+1}.
$$



En general, para cualesquiera $p$ y $d$, se tiene

$$
\Phi(B)\,\Psi(B) = \Theta(B)
$$
$$
\Longleftrightarrow 
$$
$$
\bigl(1 - \phi_1 B - \cdots - \phi_{p+d} B^{p+d}\bigr)
\bigl(1 + \psi_1 B + \psi_2 B^2 + \cdots\bigr)
=
\bigl(1 - \theta_1 B - \cdots - \theta_q B^q\bigr).
$$

Igualando coeficientes del operador \(B\), se obtiene la siguiente solución.


$$
\psi_j =
\begin{cases}
\theta_j
+ \Phi_1 \psi_{j-1}
+ \Phi_2 \psi_{j-2}
+ \cdots
+ \Phi_{p+d}\,\psi_{j-p-d},
& j = 1,\ldots,q, \\[8pt]

\Phi_1 \psi_{j-1}
+ \Phi_2 \psi_{j-2}
+ \cdots
+ \Phi_{p+d}\,\psi_{j-p-d},
& j > q,
\end{cases}
$$

$$
\text{con } \psi_0 = 1
\quad \text{y} \quad
\psi_j = 0 \;\; \text{si } j < 0.
$$


De aquí en adelante se usarán los valores estimados $\hat{\psi}_j$.

Por otra parte, dado que

$$
e_t(h) = - \sum_{j=0}^{h-1} \psi_j\, a_{t+h-j},
$$

entonces

$$
\mathbb{E}_t\!\big(e_t(h)\big) = 0,
\qquad
\operatorname{Var}_t\!\big(e_t(h)\big)
=
\sum_{j=0}^{h-1} \psi_j^{\,2}\,\sigma_a^2.
$$

Dado que $\hat{a}_t \sim N(0,\sigma_a^2)$ para todo $t$, entonces

$$
\bigl\{\, e_t(h) \mid Z_t, Z_{t-1}, \ldots, Z_1 \,\bigr\}
\sim
N\!\bigl(0,\operatorname{Var}_t(e_t(h))\bigr).
$$

Por lo tanto, los límites de $100(1-\alpha)\%$ de probabilidad para
$T(Z_{t+h})$ condicionados al conocimiento de
$Z_t, Z_{t-1}, \ldots, Z_1$
son


$$
\widehat T(Z_t)(h)
\pm
Z_{\alpha/2}
\left(
\sum_{j=0}^{h-1} \psi_j^{\,2}
\right)^{1/2}
\hat{\sigma}_a.
\qquad (7)
$$

donde $z_{\alpha/2}$ es el punto porcentual de modo que

$$
\mathbb{P}\!\left(Z > z_{\alpha/2}\right) = \frac{\alpha}{2},
\qquad
Z \sim N(0,1).
$$
